BFS guaranteed to find optimal solution. DFS not. DFS guarantees solution if finite graph with no cycles; BFS guaranteed even if infinite and cycli. 

LCFS: lowest-cost elements are removed first. I.e. use a queue (FIFO) sorted by cost; can use `heapq` in Python. Guaranteed to find optimal solution. **Pruning** only expands nodes if they have not been expanded by keeping track of already expanded end nodes. 

Best-first: remove path with the lowest **heuristic cost** $h(n)$ (might be Euclidean distance or similar). Faster than LCFS in most cases but doesn't guarantee solution. 

A-star: $f(p)=cost(node)+h(node\rightarrow goal)$. Guaranteed to find optimal solution if there is no pruning and $h$ is **admissible** (non-negative and $h(n)\leq$ cost of optimal path from $n$ to goal). To use pruning, $h$ should also be **consistent** (for $n'$ reachable from $n$, $h(n)\leq cost(n,n')+h(n')$ i.e. $f$ is non-decreasing along any path, i.e. real $cost(n,n')$ should meet or exceed reduction in heuristic from $n$ to $n'$). 

A **proposition** is a sentence with truth value, built from **atoms**, **logical connectives**, and other propositions. A **knowledge base** is a set of propositions known to be true. An **interpretation** is a function mapping all atoms in a knowledge base to true or false. A **model** is an interpretation in which all statements in the knowledge base are true. A **logical consequence** of a knowledge base is a proposition which is true in all its models.
