# Exam Notes

## Search

Nodes = **states**, arcs = **actions**. Can represent as an **explicit graph** storing all nodes and edges in memory, or as an **implicit graph** which uses some procedure to return outgoing arcs at each node, step-by-step. Use a **graph search** method to find path from start to goal node by adding paths to a **frontier** and expanding some and adding their successors using some **search strategy**. 

DFS: add starting node; add ALL edges; go down first (LIFO); add ALL edges; go down first (LIFO); etc. BFS: add starting node; add all edges; go down first (FIFO); etc.
